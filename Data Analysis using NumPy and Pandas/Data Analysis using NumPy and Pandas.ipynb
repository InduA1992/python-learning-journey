```python
# Array Operations:
```

```python
import numpy as np
```

```python
# 1. Create a 1D NumPy Array:
```

```python
temperatures_w1 = np.array([
    22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9
])

print(temperatures_w1)
```

```python
#2. Inspection and Properties:
```

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
#3. Array Operations:
```

```python
#Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32
```

```python
# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)
```

```python
# Find maximum temperature
print("Maximum temperature:", np.max(temperatures_w1))

# Find minimum temperature
print("Minimum temperature:", np.min(temperatures_w1))

# Find mean temperature
print("Mean temperature:", np.mean(temperatures_w1))
```

```python
#4. Array Slicing and Indexing:
```

```python
#Extract the temperatures for the first three days of the week.
```

```python
print("First three days:", temperatures_w1[:3])
```

```python
# Extract the temperatures for the weekend (last two days).
```

```python
print("Weekend temperatures:", temperatures_w1[-2:])
```

```python
# Extract the temperatures for the middle three days of the week.
```

```python
print("Middle three days:", temperatures_w1[1:4])
```

```python
#Pandas Series
```

```python
import pandas as pd
```

```python
#Create a Pandas Series
```

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

```python
#2. Indexing and Slicing:
```

```python
# Access the first rank using integer position 
```

```python
print("First rank mark:", marks.iloc[0])
```

```python
# Retrieve the top three ranks using labels
```

```python
print("Third rank mark:", marks.iloc[2])
```

```python
#  Filter marks greater than 90
```

```python
print("Marks greater than 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Series:
```

```python
# Change the first rank mark to 100
```

```python
marks.loc['Rank1'] = 100
print("After modifying Rank1:")
print(marks)
```

```python
# Remove the last rank
```

```python
marks = marks.drop('Rank5')
print("After removing Rank5:")
print(marks)
```

```python
# Calculate CGPA
```

```python
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
# Part 3: Pandas DataFrame
```

```python
#1. Creating Pandas DataFrame:
```

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
#2. Data Exploration:
```

```python
# Display the full DataFrame
```

```python
print(transactions)
```

```python
# First five rows
```

```python
print(transactions.head())
```

```python
# Last five rows
```

```python
print(transactions.tail())
```

```python
# Number of rows and columns
```

```python
print("Shape:", transactions.shape)
```

```python
# Column names
```

```python
print("Columns:", transactions.columns)
```

```python
# Data types and non-null counts
```

```python
transactions.info()
```

```python
# Display only ProductCategory and Amount
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Extract the last three columns using iloc
```

```python
print(transactions.iloc[:, -3:])
```

```python
# Filter North region transactions with Amount greater than 200
```

```python
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print(filtered_data)
```

```python
# Count transactions by product category
```

```python
print("Product category counts:")
print(transactions['ProductCategory'].value_counts())
```

```python
# Find unique regions
```

```python
print("Unique regions:")
print(transactions['Region'].unique())
```

```python
# Calculate mean amount by region
```

```python
print("Mean amount by region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#3. Manipulating the DataFrame:
```

```python
# Modify Amount for TransactionID 102 to 165
```

```python
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print("After modifying TransactionID 102:")
print(transactions)
```

```python
# Add a Discount column equal to 10% of Amount
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print("After adding Discount:")
print(transactions)
```

```python
# Remove the row with TransactionID 109
```

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]
print("After removing TransactionID 109:")
print(transactions)
```

```python
# Delete the Discount column
```

```python
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame:")
print(transactions)
```
